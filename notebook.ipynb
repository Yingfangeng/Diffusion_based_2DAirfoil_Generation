```python
import pickle
import pandas as pd
import os
import numpy as np

def pickle_to_csv(input_file, output_file="output.csv"):

    with open(input_file, "rb") as f:
        obj = pickle.load(f)


    if isinstance(obj, pd.DataFrame):
        df = obj
        print('The pickle file is a dataframe')

    elif isinstance(obj, (list, dict)):
        print('The pickle file is a list or dict')
        try:
            df = pd.DataFrame(obj)
            print('The pickle file can be converted into a dataframe.')
        except:
            df = pd.json_normalize(obj)
            print('The pickle file undergoes json_normalise.')

    elif isinstance(obj, np.ndarray):
        df = pd.DataFrame(obj)
        print('The pickle file is an array')

    else:
        try:
            df = pd.DataFrame(obj)
            
        except:
            raise ValueError("Cannot auto-convert this pickle content into a CSV-friendly format.")


    df.to_csv(output_file, index=False)
    print(f"Saved as: {os.path.abspath(output_file)}")


# Example usage:
pickle_to_csv("pickle_files/index_to_name_mapping.pkl", "output.csv")
```
